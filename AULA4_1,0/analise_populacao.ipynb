```python
import pandas as pd

pd.options.display.float_format = '{:,.0f}'.format
```

```python

df = pd.read_excel(
    "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx",
    sheet_name="Municípios",
    skiprows=3,      
    header=0
)

df.columns = df.columns.str.replace("\n", " ").str.strip()


df = df.dropna(subset=["UF"])

print(df.columns.tolist())
df.head()
```

```python
df = df.rename(columns={
    "NOME DO MUNICÍPIO": "MUNICIPIO",
    "População Município 2010 (Sinopse)": "POP_2010_SINOPSE",
    "População 2010 (Alterações de Limites até 2022)1": "POP_2010",
    "População Censo 2022": "POP_2022"
})

# Se a coluna POP_2010 não existir com esse nome, ajuste:
# df.columns.tolist() para conferir
df.head()
```

```python
pop_estado = (
    df.groupby("UF")[["POP_2010", "POP_2022"]]
      .sum()
      .reset_index()
)

# Diferença absoluta e crescimento percentual
pop_estado["DIFERENCA"] = pop_estado["POP_2022"] - pop_estado["POP_2010"]
pop_estado["CRESCIMENTO_%"] = (
    (pop_estado["POP_2022"] / pop_estado["POP_2010"] - 1) * 100
)

pop_estado.head()
```

```python
pop_estado = pop_estado.sort_values("DIFERENCA", ascending=False).reset_index(drop=True)

pop_estado
```

```python
pop_estado.to_csv("populacao_por_estado.csv", sep=";", index=False, encoding="utf-8-sig")
print("✅ Arquivo populacao_por_estado.csv salvo!")
```
